```python
text = """
NLP is AMAZING!!! I am learning Natural Language Processing.
I love working with text data, and NLP is very useful.
"""
```

```python
text = text.lower()
print(text)
```

```python
# remove punctuation
import string
text = text.translate(str.maketrans('','',string.punctuation))
print(text)
```

```python
import nltk
from nltk.tokenize import word_tokenize

nltk.download('punkt')
tokenize = word_tokenize(text)
print(tokenize)
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')

stop_words = set(stopwords.words('english'))
filtered = [word for word in tokenize if word not in stop_words]
print(filtered)
```

```python
texts = [
    "Win a free iPhone now",
    "Congratulations you won a prize",
    "Claim your free money",
    "Are you coming to class today",
    "Can you send me the assignment",
    "Let's meet tomorrow",
    "Please call me when you arrive",
    "Your free prize is waiting"
]

labels = [
    "spam",
    "spam",
    "spam",
    "ham",
    "ham",
    "ham",
    "ham",
    "spam"
]
```

```python
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(texts,labels,
                                                test_size = 0.2,
                                                random_state = 42
                                                )
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(ngram_range=(1,2))
X_train = vectorizer.fit_transform(X_train)
X_test = vectorizer.transform(X_test)
```

```python
from sklearn.naive_bayes import MultinomialNB
```

```python
model = MultinomialNB()
model.fit(X_train,y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix
```

```python
accuracy_score(y_test,y_pred)
```

```python
confusion_matrix(y_test,y_pred)
```

```python
from sklearn.metrics import ConfusionMatrixDisplay
import matplotlib.pyplot as plt

ConfusionMatrixDisplay.from_predictions(y_test, y_pred)

plt.show()
```
